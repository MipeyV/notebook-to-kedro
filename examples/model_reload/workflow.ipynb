```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier

df = pd.read_csv("data/training.csv")
assert not df.empty
X = df[["x"]]
y = df["y"]
```

```python
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
```

```python
predictions = model.predict(X)
fitted_model = model
```
